# Citation tracking

Add INSPIRE BibTeX keys to the lists in `categories.json`, then run the cells below. The notebook fetches current citation data from INSPIRE-HEP.

```python
import json
from pathlib import Path

from citation_tracking import CitationTracker, write_report_exports

config = json.loads(Path("categories.json").read_text(encoding="utf-8"))
report = CitationTracker(config["author_bai"]).fetch_report(config["categories"])
tex_path, text_path = write_report_exports(report, Path.cwd())
print(f"Wrote {tex_path} and {text_path}")
```

```python
def print_metrics(label, metrics):
    print(f"  {label}: {len(metrics.papers)} works | "
          f"{metrics.total_citations} citations | "
          f"h-index {metrics.h_index} | "
          f"mean {metrics.mean_citations_per_work:.2f} citations/work")

def print_papers(metrics):
    for paper in sorted(metrics.papers, key=lambda item: item.citations, reverse=True):
        print(
            f"    {paper.citations:>4} citations | {paper.author_count:>2} authors | "
            f"{'published' if paper.is_published else 'unpublished'} | "
            f"{paper.title} | {', '.join(paper.bibtex_keys)} | {paper.inspire_url}"
        )

print(f"Works on author profile: {len(report.all_papers)}")
print(f"Total citations: {report.total_citations}")
print(f"Overall h-index: {report.h_index}")
print_metrics("Published", report.published)
print_metrics("Unpublished", report.unpublished)

for label, metrics in (
    ("Short-author list (10 or fewer)", report.short_author_list),
    ("Long-author list (more than 10)", report.long_author_list),
):
    print(f"\n{label}")
    print_metrics("Summary", metrics)
    #print_papers(metrics)

for name, summary in report.categories.items():
    print(f"\n{name}")
    print_metrics("All", summary)
    print_metrics("Published", summary.published)
    print_metrics("Unpublished", summary.unpublished)
    print_metrics("Short-author list (10 or fewer)", summary.short_author_list)
    print_metrics("Long-author list (more than 10)", summary.long_author_list)
    #print_papers(summary)
```
